#### Imports and Constants

```python
main_path = "DatasetCustomImageClassification"

black_rot_path = "DatasetCustomImageClassification/BlackRot"
esca_path = "DatasetCustomImageClassification/Esca"
leaf_blight_path = "DatasetCustomImageClassification/LeafBlight"

enconded_labels_info = {
    'BlackRot' : 0,
    'Esca' : 1,
    'LeafBlight' : 2
}


kmeans_folder_path = "KMeansClusters"
kmeans_clusters_name = ["Cluster1", "Cluster2", "Cluster3"]
kmeans_clusters_sub_folder = ['BlackRot', 'Esca', 'LeafBlight']
```

```python
grayscale_image = False 
ram_limitation = False
```

```python
num_pixels = 19
image_shape = (num_pixels, num_pixels)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from PIL import Image
from PIL.ImageOps import grayscale
import cv2
import os
# Load images and labels

images = []
labels = []
paths = []

scaler = StandardScaler()

for label in ['BlackRot', 'Esca', 'LeafBlight']:
    i = 0
    if label == 'BlackRot':
        path_label = black_rot_path
    elif label == 'Esca':
        path_label = esca_path
    elif label == 'LeafBlight':
        path_label = leaf_blight_path
    
    for path in os.listdir(path_label):

        paths.append(path)

        # RAM Limitation
        #if i == 8000:
        #    print(path_label + '/' + path)
        #    break
        
        #Open image and convert to GrayScale, with the sized from histogram
        img = Image.open(path_label + '/' + path)

        if grayscale_image:
            img = grayscale(img)
            
        #resize
        img = img.resize(image_shape)

        #convert and turn 1D
        img = np.array(img).flatten()

        #Append to Array
        images.append(img)
        labels.append(label)

        i+=1

        
print("All labels done\n=============")

# Convert list to np array
images_array = np.array(images)
print(images_array.shape)
```

```python
encoded_labels = [enconded_labels_info[label] for label in labels]

print(encoded_labels)
```

```python
import umap as Umap

scaler = StandardScaler()
data_standardized = scaler.fit_transform(images_array)

umap = Umap.UMAP(n_neighbors=15, min_dist=0.1, n_components=3)
data_umap = umap.fit_transform(data_standardized, encoded_labels)
```

```python
import plotly.express as px

# Create a Pandas DataFrame for Plotly
umap_df = pd.DataFrame(data=data_umap, columns=['x', 'y', 'z'])
umap_df['Target'] = pd.Series(labels)  # Add your target variable (e.g., class labels)
umap_df['Path'] = paths

# Create a scatter plot with Plotly
fig = px.scatter_3d(umap_df, x='x', y='y', z='z', color='Target')
fig.update_layout(title='Umap Visualization',
                 xaxis_title='PC1',
                 yaxis_title='PC2')
```

#### Cluster Analysis

##### Create Folder

```python
import os

if not os.path.isdir(kmeans_folder_path):
    os.makedirs(kmeans_folder_path)

for folder_name in kmeans_clusters_name:
    if not os.path.isdir(kmeans_folder_path + "/" + folder_name):
        os.makedirs(kmeans_folder_path + "/" + folder_name)

    for sub_folder_name in kmeans_clusters_sub_folder:
        if not os.path.isdir(kmeans_folder_path + "/" + folder_name + "/" + sub_folder_name):
            os.makedirs(kmeans_folder_path + "/" + folder_name + "/" + sub_folder_name)
        
```

##### KMeans

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3)
data_kmeans = kmeans.fit_transform(data_umap, encoded_labels)

encoded_kmeans_labels = kmeans.labels_
kmeans_labels = [kmeans_clusters_name[label] for label in encoded_kmeans_labels]
```

```python
import pandas as pd

# Create a Pandas DataFrame for Plotly
kmeans_df = pd.DataFrame(data=data_kmeans, columns=['x', 'y', 'z'])
kmeans_df['Target'] = pd.Series(kmeans_labels)  # Add your target variable (e.g., class labels)
kmeans_df['TargetDataset'] = pd.Series(labels)  # Add your target variable (e.g., class labels)
kmeans_df['Path'] = paths
```

```python
# Create a scatter plot with Plotly
fig = px.scatter_3d(kmeans_df, x='x', y='y', z='z', color="Target")
fig.update_layout(title='KMeans',
                 xaxis_title='PC1',
                 yaxis_title='PC2')
```

#### Save image of each cluster

```python
import shutil

for index, row in kmeans_df.iterrows():
    for cluster_name in kmeans_clusters_name:
        if row["Target"] == cluster_name:
            shutil.copyfile(main_path + "/" + row["TargetDataset"] + "/" + row["Path"], kmeans_folder_path + "/" + cluster_name + "/" + row["TargetDataset"] + "/" + row["Path"])
            break
    
```

```python
def unique_values(seq):
    seen = set()
    seen_add = seen.add
    return [x for x in seq if not (x in seen or seen_add(x))]
```

```python
folder_sizes = []
folder_sizes_clusters = []
folder_sizes_labels = []

for cluster_name in kmeans_clusters_name:
    for label_name in kmeans_clusters_sub_folder:
        folder_sizes.append(len(os.listdir(kmeans_folder_path + "/" + cluster_name + "/" + label_name)))
        print(kmeans_folder_path + "/" + cluster_name + "/" + label_name + " : " + str(len(os.listdir(kmeans_folder_path + "/" + cluster_name + "/" + label_name))))
        folder_sizes_clusters.append(cluster_name)
        folder_sizes_labels.append(label_name)

folder_sizes = [
                [folder_sizes[0], folder_sizes[1], folder_sizes[2]],
                [folder_sizes[3], folder_sizes[4], folder_sizes[5]],
                [folder_sizes[6], folder_sizes[7], folder_sizes[8]],
                ]

display(pd.DataFrame(folder_sizes, columns=unique_values(folder_sizes_labels), index=unique_values(folder_sizes_clusters)))
```
