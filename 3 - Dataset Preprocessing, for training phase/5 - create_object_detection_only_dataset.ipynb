# Create Disease Dataset using YOLO only dataset

### <b>Objective</b> - Setting up a dataset to train yolo only for detection disease. The label identifier will be work for the CNN

#### 0 - Constants

```python
base_dir = "DatasetWithoutTrainTest/AllData/Dataset3Classes"
new_dir = "DatasetObjectDetectionOnly"

new_label = "Disease"
```

### 0.5 - Create new folders 

- DatasetObjectDetectionOnly
    - images
        - disease
            - example.JPG
            ....
    - labels
        - disease
            - example.txt
            ....
    

```python
import os

if not os.path.isdir(new_dir):
    os.mkdir(new_dir)
    os.mkdir(f'{new_dir}/images')
    os.mkdir(f'{new_dir}/images/{new_label}')
    os.mkdir(f'{new_dir}/labels')
    os.mkdir(f'{new_dir}/labels/{new_label}')
```

#### 1 - Create Dataset

```python
import pandas as pd
import shutil

list_unique_files = []

for old_label in os.listdir(f'{base_dir}/labels/'):
    for filename in os.listdir(f'{base_dir}/labels/{old_label}'):

        # Annot
        annot_filename_path = f'{base_dir}/labels/{old_label}/{filename}'
        annot_new_filename_path = f'{new_dir}/labels/{new_label}/{filename}'

        annot = pd.read_csv(annot_filename_path, sep=' ', names=["labels", "x", "y", "w", "h"])
        annot["labels"] = 0

        if not filename in list_unique_files:
            
            list_unique_files.append(filename)

            annot.to_csv(annot_new_filename_path, sep=" ", index=False, header=False)

            # Images
            image_filename_path = f'{base_dir}/images/{old_label}/{filename.replace("txt", "JPG")}'
            image_new_filename_path = f'{new_dir}/images/{new_label}/{filename.replace("txt", "JPG")}'
            shutil.copyfile(image_filename_path, image_new_filename_path)

        else:       
            
            csv_already_created = pd.read_csv(annot_new_filename_path, sep=' ', names=["labels", "x", "y", "w", "h"])

            annot = pd.concat([csv_already_created, annot], ignore_index=True)
            annot.to_csv(annot_new_filename_path, sep=" ", index=False, header=False)



       
```

#### 2 - Verify annot number

```python
import pandas as pd

new_annot = pd.DataFrame([], columns=["labels", "x", "y", "w", "h"])
base_annot = pd.DataFrame([], columns=["labels", "x", "y", "w", "h"])

for new_label in os.listdir(f'{new_dir}/labels/'):
    for filename in os.listdir(f'{new_dir}/labels/{new_label}'):

        annot_filename_path = f'{new_dir}/labels/{new_label}/{filename}'
        annot_temp = pd.read_csv(annot_filename_path, sep=' ', names=["labels", "x", "y", "w", "h"])

        new_annot = pd.concat([new_annot, annot_temp], ignore_index=True)


for old_label in os.listdir(f'{base_dir}/labels/'):
    for filename in os.listdir(f'{base_dir}/labels/{old_label}'):

        annot_filename_path = f'{base_dir}/labels/{old_label}/{filename}'
        annot_temp = pd.read_csv(annot_filename_path, sep=' ', names=["labels", "x", "y", "w", "h"])

        base_annot = pd.concat([base_annot, annot_temp], ignore_index=True)
```

```python
print(f"Original Annot Number: {base_annot.shape[0]}")
print(f"New Annot Number: {new_annot.shape[0]}")
```

#### 2.3 - Verify Image Number

Because the dataset create for only yolo have same image in different folders, we need to filter the copies

```python
image_number_1 = []
image_number_2 = []

for old_label in os.listdir(f'{base_dir}/images/'):
    image_number_1.append(([file for file in os.listdir(f'{base_dir}/images/{old_label}')]))

for new_label in os.listdir(f'{new_dir}/images/'):
    image_number_2.append(([file for file in os.listdir(f'{new_dir}/images/{new_label}')]))

image_number_1 = list(set([item for sublist in image_number_1 for item in sublist]))
image_number_2 = [item for sublist in image_number_2 for item in sublist]

print(f"Original Images Number: {len(image_number_1)}")
print(f"New Images Number: {len(image_number_2)}")
```
