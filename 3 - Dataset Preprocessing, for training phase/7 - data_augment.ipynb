### Data Augment CNN

1. Copy to new folder
2. Add augmentation to 40000 images per class
3. Save new image to folder

```python
import os
import shutil

base_folder = "DatasetClassification"
new_folder = "DatasetClassificationWithDataAugment"

if not os.path.isdir(new_folder):
    shutil.copytree(base_folder, new_folder, dirs_exist_ok=True)
```

```python
import os

dataset_path = "DatasetClassificationWithDataAugment"
labels = ["BlackRot", "Esca", "LeafBlight"]
label_paths = [] 

print("Image number:")

i = 0
for label in labels:
    label_paths.append(os.listdir(os.path.join(dataset_path, label)))
    print(f"{label}: {len(os.listdir(os.path.join(dataset_path, label)))}")
```

```python
# Image Classification
import torch
import torchvision
from torchvision.transforms import v2
from PIL import Image

v2.RandAugment()

transforms = v2.Compose([
    v2.ToTensor(),
    v2.ToDtype(torch.float32, scale=True),
    v2.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    #v2.RandAugment()
    v2.ToPILImage()
])

max_value = 40000

i = 0
for label in labels:
    number_of_images_augmented_class = max_value - len(label_paths[i]) 
    count = 0
    for img_name in label_paths[i]:

        if count >= number_of_images_augmented_class:
            continue

        path = os.path.join(str(dataset_path), str(label), str(img_name))
        #print(dataset_path)
        #print(label)
        #print(img_name)
        img = Image.open(path)

        img = transforms(img)

        #display(img)

        new_name = path.replace(".JPG", "") + "_DataAugmented.JPG"
        img.save(new_name)

        count+=1

    print(f"{label}: {count} with total of {count + len(label_paths[i])}")

    i+=1
```
