# Mix the training and test data

```python
import os
import shutil

base_folder = "DatasetOrganized"
new_folder = "DatasetWithoutTrainTest"

if not os.path.isdir(new_folder):
    shutil.copytree(base_folder, new_folder, dirs_exist_ok=True)
```

```python
import os
import shutil

main_path = new_folder

for dataset_type in os.listdir(main_path):
    for dataset in os.listdir(main_path + "/" + dataset_type):
        for im_lb in os.listdir(main_path + "/" + dataset_type + "/" + dataset):
            for tr in ["train", "test"]:
                for label in os.listdir(main_path + "/" + dataset_type + "/" + dataset + "/" + im_lb + "/" + tr):                  
                    new_path = main_path + "/" + dataset_type + "/" + dataset + "/" + im_lb + "/" + label

                    if not os.path.isdir(new_path):
                        os.mkdir(new_path)
                                       
                    for file in os.listdir(main_path + "/" + dataset_type + "/" + dataset + "/" + im_lb + "/" + tr + "/" + label):
                        old_path = main_path + "/" + dataset_type + "/" + dataset + "/" + im_lb + "/" + tr + "/" + label + "/" + file
                        new_path_file = new_path + "/" + file
                        shutil.move(old_path, new_path_file)
                    
                shutil.rmtree(main_path + "/" + dataset_type + "/" + dataset + "/" + im_lb + "/" + tr)

            
```
