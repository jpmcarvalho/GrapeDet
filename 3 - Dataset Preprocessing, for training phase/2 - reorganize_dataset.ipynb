# Reorganize Dataset

#### <b>Objective</b>: Some Labels are out of place (BlackRot label in Esca Folder). Most Important for multiclass, since the first notebook separate the annotation between folders, without changing the original one

<!-->

#### 0 - Constants

```python
labels = ["BlackRot", "Esca", "LeafBlight"]

old_directory = "Dataset"
new_directory = "DatasetOrganized"
```

#### 0.5 Create Folders with the same structure of the original

```python
import os
import shutil


if not os.path.isdir("DatasetOrganized"):
    os.mkdir("DatasetOrganized")

    for dir in os.listdir(old_directory):
        if not "." in dir:
            os.mkdir(new_directory + "/" + dir)
        else:
            continue

        for sub_dir_1 in os.listdir(old_directory + "/" + dir):
            if not "." in sub_dir_1:
                os.mkdir(new_directory + "/" + dir + "/" + sub_dir_1)
            else:
                continue

            for sub_dir_2 in os.listdir(old_directory + "/" + dir + "/" + sub_dir_1):
                if not "." in sub_dir_2:
                    os.mkdir(new_directory + "/" + dir + "/" + sub_dir_1 + "/" + sub_dir_2)
                else:
                    continue

                for sub_dir_3 in os.listdir(old_directory + "/" + dir + "/" + sub_dir_1 + "/" + sub_dir_2):
                    if not "." in sub_dir_3:
                        os.mkdir(new_directory + "/" + dir + "/" + sub_dir_1 + "/" + sub_dir_2 + "/" + sub_dir_3)
                    else:
                        continue

                    for sub_dir_4 in os.listdir(old_directory + "/" + dir + "/" + sub_dir_1 + "/" + sub_dir_2 + "/" + sub_dir_3):
                        if not "." in sub_dir_4:
                            os.mkdir(new_directory + "/" + dir + "/" + sub_dir_1 + "/" + sub_dir_2 + "/" + sub_dir_3 + "/" + sub_dir_4)
                        else:
                            continue
```

#### 1 - Send data to each folder (Uniclass)

##### 1.1 - BlackRot Dataset

```python
import pandas as pd

label = "BlackRot"


for dir in os.listdir(old_directory):
    if "." in dir:
        continue           

    for sub_dir_1 in os.listdir(old_directory + "/" + dir):
        if "." in sub_dir_1 or not label in sub_dir_1:
            continue

        for sub_dir_2 in os.listdir(old_directory + "/" + dir + "/" + sub_dir_1):
            if "." in sub_dir_2 or "images" in sub_dir_2:
                continue


            for sub_dir_3 in os.listdir(old_directory + "/" + dir + "/" + sub_dir_1 + "/" + sub_dir_2):
                if "." in sub_dir_3:
                    continue

                labels_dir = [labels_dir for labels_dir in os.listdir(old_directory + "/" + dir + "/" + sub_dir_1 + "/" + sub_dir_2 + "/" + sub_dir_3)]

                for sub_dir_4 in os.listdir(old_directory + "/" + dir + "/" + sub_dir_1 + "/" + sub_dir_2 + "/" + sub_dir_3):
                    if "." in sub_dir_4:
                        continue
            
                    old_path = old_directory + "/" + dir + "/" + sub_dir_1 + "/" + sub_dir_2 + "/" + sub_dir_3 + "/" + sub_dir_4
                    new_path = new_directory + "/" + dir + "/" + sub_dir_1 + "/" + sub_dir_2 + "/" + sub_dir_3 + "/"
                
                    for file_name in os.listdir(old_path):
                        # Verify if file is empty
                        if os.path.getsize(old_path + "/" + file_name) == 0:
                            print("File " + file_name + " is empty")
                            continue

                        annot_temp = pd.read_csv(old_path + "/" + file_name, header=None, sep=" ")
                        annot_temp.columns = ["label", "x", "y", "w", "h"]

                        # Create an pd for blackrot
                        annot_br = annot_temp
                        if annot_br.shape[0] > 0:
                            annot_br.to_csv(new_path + labels_dir[0] + "/" + file_name, sep=" ", header=False, index=False)
                            shutil.copyfile(
                                old_path.replace("labels", "images") + "/" + file_name.replace("txt", "JPG"), 
                                new_path.replace("labels", "images") + labels_dir[0] + "/" + file_name.replace("txt", "JPG"),
                            )
```

```python
import pandas as pd

label = "3Classes"


for dir in os.listdir(old_directory):
    if "." in dir:
        continue           

    for sub_dir_1 in os.listdir(old_directory + "/" + dir):
        if "." in sub_dir_1 or not label in sub_dir_1:
            continue

        for sub_dir_2 in os.listdir(old_directory + "/" + dir + "/" + sub_dir_1):
            if "." in sub_dir_2 or "images" in sub_dir_2:
                continue


            for sub_dir_3 in os.listdir(old_directory + "/" + dir + "/" + sub_dir_1 + "/" + sub_dir_2):
                if "." in sub_dir_3:
                    continue

                labels_dir = [labels_dir for labels_dir in os.listdir(old_directory + "/" + dir + "/" + sub_dir_1 + "/" + sub_dir_2 + "/" + sub_dir_3)]

                for sub_dir_4 in os.listdir(old_directory + "/" + dir + "/" + sub_dir_1 + "/" + sub_dir_2 + "/" + sub_dir_3):
                    if "." in sub_dir_4:
                        continue
            
                    old_path = old_directory + "/" + dir + "/" + sub_dir_1 + "/" + sub_dir_2 + "/" + sub_dir_3 + "/" + sub_dir_4
                    new_path = new_directory + "/" + dir + "/" + sub_dir_1 + "/" + sub_dir_2 + "/" + sub_dir_3 + "/"

                
                    for file_name in os.listdir(old_path):
                        
                        # Verify if file is empty
                        if os.path.getsize(old_path + "/" + file_name) == 0:
                            print("File " + file_name + " is empty")
                            continue

                        annot_temp = pd.read_csv(old_path + "/" + file_name, header=None, sep=" ")
                        annot_temp.columns = ["label", "x", "y", "w", "h"]

                        # Create an pd for blackrot
                        annot_br = annot_temp[annot_temp["label"] == 0]
                        if annot_br.shape[0] > 0:
                            annot_br.to_csv(new_path + labels_dir[0] + "/" + file_name, sep=" ", header=False, index=False)
                            shutil.copyfile(
                                old_path.replace("labels", "images") + "/" + file_name.replace("txt", "JPG"), 
                                new_path.replace("labels", "images") + labels_dir[0] + "/" + file_name.replace("txt", "JPG"),
                            )

                        # Create an pd for blackrot
                        annot_e = annot_temp[annot_temp["label"] == 1]
                        if annot_e.shape[0] > 0:
                            annot_e.to_csv(new_path + labels_dir[1] + "/" + file_name, sep=" ", header=False, index=False)
                            shutil.copyfile(
                                old_path.replace("labels", "images") + "/" + file_name.replace("txt", "JPG"), 
                                new_path.replace("labels", "images") + labels_dir[1] + "/" + file_name.replace("txt", "JPG"),
                            )

                        # Create an pd for blackrot
                        annot_lb = annot_temp[annot_temp["label"] == 2]
                        if annot_lb.shape[0] > 0:
                            annot_lb.to_csv(new_path + labels_dir[2] + "/" + file_name, sep=" ", header=False, index=False)
                            shutil.copyfile(
                                old_path.replace("labels", "images") + "/" + file_name.replace("txt", "JPG"), 
                                new_path.replace("labels", "images") + labels_dir[2] + "/" + file_name.replace("txt", "JPG"),
                            )
```
