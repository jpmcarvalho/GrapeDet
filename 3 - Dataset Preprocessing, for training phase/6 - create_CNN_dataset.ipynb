## Create Dataset for Classification Training, using the manual annotation of YOLO

### <b>Objective</b> - Setting up a dataset to train yolo only for detection disease. The label identifier will be work for the Classification Only Task

```python
import os

base_dir = "DatasetWithoutTrainTest/AllData/Dataset3Classes"
new_dir = "DatasetClassification"

# Get the label name
label = next(os.walk(f"{base_dir}/images"))[1][0]
```

```python
import os

if not os.path.isdir(new_dir):
    os.mkdir(new_dir)

    for folder in os.listdir(f'{base_dir}/labels/'):
        os.mkdir(f'{new_dir}/{folder}')
```

#### Create dataset

```python
import pandas as pd
import numpy as np
import shutil
from PIL import Image
import cv2

list_unique_files = []

for folder in os.listdir(f'{base_dir}/labels/'):
    for filename in os.listdir(f'{base_dir}/labels/{folder}'):
        
        # Annot
        annot_filename_path = f'{base_dir}/labels/{folder}/{filename}'
        image_path = f'{base_dir}/images/{folder}/{filename.replace("txt", "JPG")}'

        cropped_image_path = f'{new_dir}/{folder}/{filename.replace("txt", "JPG")}'

        annot = pd.read_csv(annot_filename_path, sep=' ', names=["labels", "x", "y", "w", "h"])

        # Cut image with annot area
        for index, row in annot.iterrows():
            img = Image.open(image_path)
            open_cv_image = np.array(img)

            image_bgr = cv2.cvtColor(open_cv_image, cv2.COLOR_RGB2BGR)

            X = int(row['x'] * image_bgr.shape[0])
            Y = int(row['y'] * image_bgr.shape[1])
            W = int(row['w'] * image_bgr.shape[0])
            H = int(row['h'] * image_bgr.shape[1])

            cropped_image = image_bgr[int(Y-H/2):int(Y+H/2), int(X-W/2):int(X+W/2)]
            crop_img = Image.fromarray(cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB))
            crop_img.save(f"{cropped_image_path}_{index}.JPG")
```

```python
sum = 0
for folder in os.listdir(f'{new_dir}'):
    sum += len(os.listdir(f"{new_dir}/{folder}"))
    print(len(os.listdir(f"{new_dir}/{folder}")))

print(f"Sum: {sum}")
```
