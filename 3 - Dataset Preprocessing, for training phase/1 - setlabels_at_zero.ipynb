# Set Labels of the class as 0 for uniclass YOLO

#### <b>Objective</b>: Filter and process content of multiclass dataset for uniclass dataset

#### 0 - Create Directories

```python
import os

directory = "Dataset/AllData"
sub_folders = ["DatasetBlackRot", "DatasetEsca", "DatasetLeafBlight"]

if not os.path.isdir(directory):
    os.mkdir(directory)

for sb in sub_folders:
    os.mkdir(directory + "/" + sb)
    for sb1 in ['images', 'labels']:
        os.mkdir(directory + "/" + sb + "/" + sb1)

        for sb2 in ['train', 'test']:
            os.mkdir(directory + "/" + sb + "/" + sb1 + "/" + sb2)

            if "BlackRot" in sb:
                os.mkdir(directory + "/" + sb + "/" + sb1 + "/" + sb2 + "/" + "BlackRot")

            elif "Esca" in sb:
                os.mkdir(directory + "/" + sb + "/" + sb1 + "/" + sb2 + "/" + "Esca")

            elif "LeafBlight" in sb:
                os.mkdir(directory + "/" + sb + "/" + sb1 + "/" + sb2 + "/" + "LeafBlight")

```

#### 0.5 - Setup Paths

```python
old_directory = "Dataset/AllData/Dataset3Classes"
black_rot_directory = "Dataset/AllData/DatasetBlackRot"
esca_directory = "Dataset/AllData/DatasetEsca"
leaf_blight_directory = "Dataset/AllData/DatasetLeafBlight"
```

### 1 - Dataset BlackRot

```python
import shutil
import pandas as pd

label = "BlackRot"

for sb2 in ['train', 'test']:
        
    # Copy updated annot
    for filename in os.listdir(old_directory + "/labels/" + sb2 + "/" + label):

        old_annot_path = old_directory + "/labels/" + sb2 + "/" + label + "/" + filename
        new_annot_path = black_rot_directory + "/labels/" + sb2 + "/" + label + "/" + filename


        annot_temp = pd.read_csv(old_directory + "/labels/" + sb2 + "/" + label + "/" + filename, header=None, sep=" ")
        annot_temp.columns = ["label", "x", "y", "w", "h"]
        

        annot_black_rot = annot_temp[annot_temp["label"] == 0]
        annot_esca = annot_temp[annot_temp["label"] == 1]
        annot_leaf_blight = annot_temp[annot_temp["label"] == 2]

        # If it have BlackRot
        if annot_black_rot.shape[0] > 0:

            #Labels
            new_annot_path = black_rot_directory + "/labels/" + sb2 + "/BlackRot/" + filename
            annot_black_rot["label"] = 0
            annot_black_rot.to_csv(new_annot_path, sep=" ", header=False, index=False) 

            #Image
            old_image_path = old_directory + "/images/" + sb2 + "/" + label + "/" + filename.replace("txt", "JPG")
            new_image_path = black_rot_directory + "/images/" + sb2 + "/BlackRot/" + filename.replace("txt", "JPG")
            shutil.copyfile(old_image_path, new_image_path)

        # If it have Esca
        if annot_esca.shape[0] > 0:

            #Labels
            new_annot_path = esca_directory + "/labels/" + sb2 + "/Esca/" + filename
            annot_esca["label"] = 0
            annot_esca.to_csv(new_annot_path, sep=" ", header=False, index=False)

            #Image
            old_image_path = old_directory + "/images/" + sb2 + "/" + label + "/" + filename.replace("txt", "JPG")
            new_image_path = esca_directory + "/images/" + sb2 + "/Esca/" + filename.replace("txt", "JPG")
            shutil.copyfile(old_image_path, new_image_path)

        # If it have LeafBlight
        if annot_leaf_blight.shape[0] > 0:

            #Labels
            new_annot_path = leaf_blight_directory + "/labels/" + sb2 + "/LeafBlight/" + filename
            annot_leaf_blight["label"] = 0
            annot_leaf_blight.to_csv(new_annot_path, sep=" ", header=False, index=False)

            #Image
            old_image_path = old_directory + "/images/" + sb2 + "/" + label + "/" + filename.replace("txt", "JPG")
            new_image_path = leaf_blight_directory + "/images/" + sb2 + "/LeafBlight/" + filename.replace("txt", "JPG")
            shutil.copyfile(old_image_path, new_image_path)
        
```

```python
import pandas as pd
import os

annot_br = pd.DataFrame([], columns=["x", "y", "w", "h"])

for folder in ["train", "test"]:
    for filename in os.listdir("Dataset/AllData/DatasetBlackRot/labels/" + folder + "/BlackRot"):
        annot_temp = pd.read_csv("Dataset/AllData/DatasetBlackRot/labels/" + folder + "/BlackRot/" + filename, header=None, sep=" ")
        annot_br = pd.concat([annot_br, annot_temp], ignore_index=True)

print("Total of BlackRot annotation: " + str(annot_br.shape[0]))
```
